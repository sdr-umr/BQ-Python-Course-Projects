# Project 1: Simple Calculator

```python
class Calculator:

    def __init__(self, a, b):
        self.a = a
        self.b = b

    def calculate(self):
        self.add = self.a + self.b
        self.sub = self.a - self.b
        self.mul = self.a * self.b
        self.div = self.a / self.b

    def display(self):
        print("Addition:", self.add)
        print("Subtraction:", self.sub)
        print("Multiplication:", self.mul)
        print("Division:", self.div)


a = float(input("First number: "))
b = float(input("Second number: "))

calc = Calculator(a, b)
calc.calculate()
calc.display()
        
```

# #Project 2: Temperature Converter

```python
class Temperature:

    def __init__(self, value):
        self.value = value

    def c_to_f(self):
        return (self.value * 9 / 5) + 32

    def f_to_c(self):
        return (self.value - 32) * 5 / 9


c = float(input("Enter Celsius value to convert in Farehnheit: "))
obj = Temperature(c)
print("Celsius to Fahrenheit is:", obj.c_to_f())

f = float(input("Enter Fahrenheit value to convert in Celsius: "))
obj = Temperature(f)
print("Farehnheit to Celsius is:", obj.f_to_c())
```

# Project 3: Menu Driven Calculator

```python
class Calculator:

    def __init__(self, a, b):
        self.a = a
        self.b = b

    def calculate(self, choice):
        if choice == "+":
            return self.a + self.b
        elif choice == "-":
            return self.a - self.b
        elif choice == "*":
            return self.a * self.b
        elif choice == "/":
            if self.b == 0:
                return "Division by zero is not possible."
            return self.a / self.b


while True:
    print("1: + Addition")
    print("2: - Subtraction")
    print("3: * Multiplication")
    print("4: / Division")
    print("5: 0 Exit")

    choice = input("What operation you want to Perform? Enter the Symbol of Operation(+-*/): ")

    if choice == "0":
        print("Calculator closed.")
        break

    if choice not in "+-*/":
        print("Invalid choice.")
        continue

    a = float(input("First number: "))
    b = float(input("Second number: "))

    calc = Calculator(a, b)
    print("Answer:", calc.calculate(choice))
```
